```python
import pandas as pd
import numpy as np

# Hacer listas con los datos de pastelerías en Guadalajara
datos_pastelerias = {
    'Pasteleria': [
        'Marisa','Marisa','Marisa',
        'Petit Repostería con Alma','Petit Repostería con Alma','Petit Repostería con Alma',
        'Ok Pastelería','OK Pastelería','OK Pastelería',
        'Pastelería fina Neufeld','Pastelería fina Neufeld','Pastelería fina Neufeld'
         ],

    'Producto_Competencia': [
        'Pastel de Moka', 'Pastel Tres Leches','Pastel de zanahoria',
        'Moka','Tres leches con nuez','Zanahoria',
        'Pastel de Moka','Pastel Tres Leches','Pastel de Zanahoria',
        'Pastel Red Moka','Pastel 3 leches nuez','Pastel Red Zanahoria'
        ],
    'Categoria_Producto':[
        'Moka','Tres Leches','Zanahoria',
        'Moka','Tres Leches','Zanahoria',
        'Moka','Tres Leches','Zanahoria',
        'Moka','Tres Leches','Zanahoria'
    ],
    'Precio_Venta_MXN': [
        395.00,235.00,230.00,
        519.00,439.00,494.00,
        478.00,454.00,485.00,
        470.00,470.00,490.00
    ]
}

# Construir el dataframe
df_smartmenu = pd.DataFrame(datos_pastelerias)

# Calcular el promedio específico de cada categoría
# Esto evita mezclar las peras con las manzanas (no compara el precio de un moka con un tres leches)
df_smartmenu['Promedio_Categoria_MXN'] = df_smartmenu.groupby('Categoria_Producto')['Precio_Venta_MXN'].transform('mean').round(2)

# Calcular la desviación exacta contra su propia categoría
df_smartmenu['Desviacion_Categoria_MXN'] = df_smartmenu['Precio_Venta_MXN'] - df_smartmenu['Promedio_Categoria_MXN']

# Clasificar de manera automatizada de posicionamiento comercial
df_smartmenu['Posicionamiento_Mercado'] = np.where(
    df_smartmenu['Desviacion_Categoria_MXN'] < 0,
    'Estrategia de Volumen / Precio Competitivo',
    'Posicionamiento Premium / Enfoque en Margen'
)

# Exportar la nueva matriz robusta a CSV
df_smartmenu.to_csv('smartmenu_gdl_dataset.csv', index=False)
print("=== SMARTMENU ANALYTICS: PIPELINE DE DATOS COMPLETO ===")
print(f"Se ha generado una matriz con {len(df_smartmenu)} registros competitivos.")
print("Archivo 'smartmenu_gdl_dataset.csv' actualizado de forma local. 🟢\n")

# Mostrar el resultado
df_smartmenu
```
